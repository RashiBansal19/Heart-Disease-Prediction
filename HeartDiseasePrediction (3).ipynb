Importing the Dependencies

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

Data Collection and Processing


```python
# loading the csv data to a Pandas DataFrame
heart_data = pd.read_csv('/content/HeartDiseaseTrain-Test.csv')
```

```python
# print first 5 rows of the dataset
heart_data.head()
```

```python
# print last 5 rows of the dataset
heart_data.tail()
```

```python
# number of rows and columns in the dataset
heart_data.shape
```

```python
# getting some info about the data
heart_data.info()
```

```python
# checking for missing values
heart_data.isnull().sum()
```

```python
# statistical measures about the data
heart_data.describe()
```

```python
# checking the distribution of Target Variable
heart_data['target'].value_counts()
```

```python
heart_data = pd.get_dummies(heart_data, drop_first=True)
```

```python
original_columns = heart_data.drop('target', axis=1).columns
```

1 --> Defective Heart
0 --> Healthy Heart

Splitting the Features and Target

```python
X = heart_data.drop(columns='target', axis=1)   # whenever dropping column axis is 1 in row it is 0
Y = heart_data['target']
```

```python
print(X)
```

```python
print(Y)
```

Splitting the Data into Training data & Test Data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=2)
# test_size=0.2 takes 20 percent data as testing data
# stratify distributes target evenly between testing and training data
```

```python
print(X.shape, X_train.shape, X_test.shape)
# 80 percent data in training data and 20 percent in testing
```

Model Training

Logistic Regression

```python
model = LogisticRegression()

# training the LogisticRegression model with Training data
model.fit(X_train, Y_train)
```

Model Evaluation

Accuracy Score

```python
# accuracy on training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
```

```python
print('Accuracy on Training data : ', training_data_accuracy)
```

```python
# accuracy on test data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
```

```python
print('Accuracy on Testing data : ', test_data_accuracy)
```

Building a Predictive System

```python
input_data = {
    'age': 41,
    'sex': 'Male',
    'chest_pain': 'Atypical angina',
    'resting_bp': 130,
    'cholesterol': 204,
    'fasting_blood_sugar': 'Lower than 120 mg/dl',
    'resting_ecg': 'Normal',
    'Max_heart_rate': 172,
    'exercise_induced_angina': 'No',
    'oldpeak': 1.4,
    'slope': 'Downsloping',
    'vessels_col': 'Zero',
    'thalassemia': 'Reversible Defect'
}

input_df = pd.DataFrame([input_data])

# Apply exactly the same encoding used for training
input_df = pd.get_dummies(input_df, drop_first=True)

# Make the input have exactly the same 22 columns as X
input_df = input_df.reindex(columns=X.columns, fill_value=False)

# Prediction
prediction = model.predict(input_df)

print(prediction)
```

```python
if (prediction[0] == 0):
    print('The Person does not have a Heart Disease')
else:
    print('The Person has Heart Disease')
```
